```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

np.random.seed(42)
random.seed(42)

# --- Config ---
NUM_USERS = 2000
START_DATE = datetime(2024, 1, 1)
END_DATE   = datetime(2024, 6, 30)

FEATURES = [
    'login', 'dashboard_view', 'report_create',
    'report_share', 'export_data', 'invite_teammate',
    'settings_change', 'api_access', 'billing_view', 'logout'
]

PLANS = ['free', 'starter', 'pro', 'enterprise']
PLAN_WEIGHTS = [0.50, 0.25, 0.15, 0.10]

# --- Generate Users ---
users = pd.DataFrame({
    'user_id': [f'U{str(i).zfill(4)}' for i in range(1, NUM_USERS + 1)],
    'signup_date': [START_DATE + timedelta(days=random.randint(0, 90)) for _ in range(NUM_USERS)],
    'plan': random.choices(PLANS, weights=PLAN_WEIGHTS, k=NUM_USERS),
    'country': random.choices(['India','USA','UK','Germany','Singapore'], 
                               weights=[0.4,0.3,0.1,0.1,0.1], k=NUM_USERS)
})

# --- Generate Events ---
records = []
for _, user in users.iterrows():
    # Active users generate more events; churned users stop early
    is_churned = random.random() < 0.35
    churn_day  = random.randint(10, 60) if is_churned else 999
    
    num_events = random.randint(5, 120)
    for _ in range(num_events):
        day_offset = random.randint(0, (END_DATE - user['signup_date']).days)
        if day_offset > churn_day:
            continue
        event_date = user['signup_date'] + timedelta(days=day_offset)
        if event_date > END_DATE:
            continue
        records.append({
            'user_id':    user['user_id'],
            'event_date': event_date.date(),
            'event_name': random.choices(
                FEATURES,
                weights=[20,15,12,8,8,5,5,5,4,18],
                k=1
            )[0],
            'plan':       user['plan'],
            'country':    user['country'],
            'signup_date': user['signup_date'].date()
        })

df = pd.DataFrame(records)
df.to_csv(r"C:\Users\Soumili Nag\OneDrive\Desktop\projects\SaaS Product Analytics\data\saas_events.csv", index=False)
print(f"Dataset created: {df.shape[0]} rows, {df['user_id'].nunique()} unique users")
df.head()
```
